# Bloque 2 Definir el dato que medimos 

```python
temperatura = 36.0
print("Temperatura registrada: ", temperatura, " C*")

humedad = 60.0
print("Humedad relativa: ", humedad,"%")
```

# Bloque 3 Definir los conjuntos [a,b,c]

```python
mediaT = [20, 30, 40]
altaT = [30, 40, 50]

print("Categoria media: ", mediaT)
print("Categoria alta: ", altaT)

altaH = [50, 70, 100]
print("Categoria alta: ",altaH)
```

# Bloque 4 Aplicar formulas

```python
a = 20
b = 30
c = 40
pertenencia_media_t = (c - temperatura) / (c - b)
print("Grado de pertenencia a media: ", pertenencia_media_t)
```

# Bloque 5 Funcion triangular

```python
def pertenencia_triangular(x, puntos):
    a = puntos[0]
    b = puntos[1]
    c = puntos[2]

    if x <= a or x >= c:
        return 0.0
    elif x == b: 
        return 1.0
    elif x < b:
        return (x - a) / (b - a)
    else: 
        return (c - x) / (c - b)

print("Funcion de pertenencia triangular")
```

# Bloque 6 Calcular los grados para temperatura y humedad

```python
grado_cat_media_t = pertenencia_triangular(temperatura, mediaT)
grado_cat_alta_t = pertenencia_triangular(temperatura, altaT)

print("Temperatura: ", temperatura)
print("Grado de pertenencia de la categoria Media: ", grado_cat_media_t)
print("Grado de pertenencia de la categoria Alta: ", grado_cat_alta_t)

grado_cat_alta_h = pertenencia_triangular(humedad, altaH)
print("Humedad: ", humedad)
print("Grado de pertenencia de la categoria Alta: ", grado_cat_alta_h)
```

# Bloque 7 Pertenencias del triangulos

```python
# Temperatura
print("Temperatura media de 30 grados: ", pertenencia_triangular(30, mediaT))
print("Temperatura alta de 50 grados: ", pertenencia_triangular(40, altaT))
# Humedad
print("Humedad alta del 60%: ", pertenencia_triangular(60, altaH))
```
