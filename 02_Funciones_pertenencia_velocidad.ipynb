# Bloque 1 Importar librerias

```python
import numpy as np
import matplotlib.pyplot as plt
print("Listo")
```

# Bloque 2 Definir el dato que medimos

```python
velocidad = 35.0
print("Velocidad registrada: ", velocidad, " km/h")
```

# Bloque 3 Definir los 3 conjuntos [a,b,c]

```python
lenta = [0, 20, 40]
media = [30, 50, 70]
rapida = [60, 80, 100]

print("Categoria lenta: ", lenta)
print("Categoria media: ", media)
print("Categoria rapida: ", rapida)
```

# Bloque 4 Aplicar formulas
 Subida = (x - a) / (b - a)
 Bajada = (c - x) / (c - b)

```python
a = 30
b = 50 
c = 70
pertenencia_media = (c - velocidad) / (c - b)
print("Grado de pertenencia a MEDIA: ", pertenencia_media)
```

# Bloque 5 Funcion triangular

```python
def pertenencia_triangular(x, puntos):
    a = puntos[0]
    b = puntos[1]
    c = puntos[2]

    if x <= a or x >= c:
        return 0.0
    elif x == b: 
        return 1.0
    elif x < b:
        return (x - a) / (b - a)
    else: 
        return (c - x) / (c - b)

print("Funcion de pertenencia triangular")
```

# Bloque 6 Calcular los tres grados para los 35km/h

```python
grado_cat_lenta = pertenencia_triangular(velocidad, lenta)
grado_cat_media = pertenencia_triangular(velocidad, media)
grado_cat_rapida = pertenencia_triangular(velocidad, rapida)

print("Velocidad: ", velocidad)
print("Grado de pertenencia de la categoria Baja: ",grado_cat_lenta)
print("Grado de pertenencia de la categoria Media: ",grado_cat_media)
print("Grado de pertenencia de la categoria Alta: ",grado_cat_rapida)
```

# Bloque 7 Tres velocidades distintas, una en el pico de cada triangulo

```python
print("Lenta, 30 km/h: ", pertenencia_triangular(30, lenta))
print("Media, 50 km/h: ", pertenencia_triangular(50, media))
print("Rapida, 70 km/h: ", pertenencia_triangular(70, rapida))
```

# Bloque 8 Dibujar las funciones de pertenencia

```python
velocidad_para_grafica = np.linspace(0,100,401)

categoria_lenta = []
categoria_media = []
categoria_rapida = []

for valor in velocidad_para_grafica: 
    categoria_lenta.append(pertenencia_triangular(valor,lenta))
    categoria_media.append(pertenencia_triangular(valor,media))
    categoria_rapida.append(pertenencia_triangular(valor,rapida))

print("Listo")
```

# Bloque 9 Mostrar la grafica de las tres categorias

```python
plt.figure(figsize=(10, 5))

plt.plot(velocidad_para_grafica, categoria_lenta, label = "Lenta", linewidth = 2)
plt.plot(velocidad_para_grafica, categoria_media, label = "Media", linewidth = 2)
plt.plot(velocidad_para_grafica, categoria_rapida, label = "Rapida", linewidth = 2)

plt.axvline(velocidad, linestyle="--", color="black", label="Dato: 35 km/h")
plt.xlabel("Velocidad (km/h)")
plt.ylabel("Grado de pertenencia (0 a 1)")
plt.title("Tres conjuntos difusos para una sola variable")
plt.ylim(-0.05, 1.05)
plt.grid(alpha=0.3)
plt.legend()
plt.show()
```
