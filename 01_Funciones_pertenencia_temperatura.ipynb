# Bloque 1 Importar librerias

```python
import numpy as np
import matplotlib.pyplot as plt
print("Listo")
```

# Bloque 2 Definir el dato que medimos

```python
temperatura = 35.0
print("Temperatura registrada: ", temperatura, " C*")
```

# Bloque 3 Definir los 3 conjuntos [a,b,c]

```python
baja = [10, 20, 30]
media = [20, 30, 40]
alta = [30, 40, 50]

print("Categoria baja: ", baja)
print("Categoria media: ", media)
print("Categoria alta: ", alta)
```

# Bloque 4 Aplicar formulas
 Subida = (x - a) / (b - a)
 Bajada = (c - x) / (c - b)

```python
a = 20
b = 30
c = 40
pertenencia_media = (c - temperatura) / (c - b)
print("Grado de pertenencia a MEDIA: ", pertenencia_media)
```

# Bloque 5 Funcion triangular

```python
def pertenencia_triangular(x, puntos):
    a = puntos[0]
    b = puntos[1]
    c = puntos[2]

    if x <= a or x >= c:
        return 0.0
    elif x == b: 
        return 1.0
    elif x < b:
        return (x - a) / (b - a)
    else: 
        return (c - x) / (c - b)

print("Funcion de pertenencia triangular")
```

# Bloque 6 Calcular los tres grados para el 35*

```python
grado_cat_baja = pertenencia_triangular(temperatura, baja)
grado_cat_media = pertenencia_triangular(temperatura, media)
grado_cat_alta = pertenencia_triangular(temperatura, alta)

print("Temperatura: ", temperatura)
print("Grado de pertenencia de la categoria Baja: ",grado_cat_baja)
print("Grado de pertenencia de la categoria Media: ",grado_cat_media)
print("Grado de pertenencia de la categoria Alta: ",grado_cat_alta)
```

# Bloque 7 Tres temperaturas distintas, una en el pico de cada triangulo

```python
print("Baja, 20 grados: ", pertenencia_triangular(20, baja))
print("Media, 30 grados: ", pertenencia_triangular(30, media))
print("Alta, 40 grados: ", pertenencia_triangular(40, alta))
```

# Bloque 8 ¿Cuanto pertenece 25*C a cada conjunto?

```python
temperatura_25grados = 25.0

print("Baja: ",pertenencia_triangular(temperatura_25grados, baja))
print("Media: ", pertenencia_triangular(temperatura_25grados, media))
print("Alta: ", pertenencia_triangular(temperatura_25grados, alta))
```

# Bloque 9 Dibujar las funciones de pertenencia

```python
temperatura_para_grafica = np.linspace(10,50,401)

categoria_baja = []
categoria_media = []
categoria_alta = []

for valor in temperatura_para_grafica: 
    categoria_baja.append(pertenencia_triangular(valor,baja))
    categoria_media.append(pertenencia_triangular(valor,media))
    categoria_alta.append(pertenencia_triangular(valor,alta))

print("Listo")
```

# Bloque 10 Mostrar la grafica de las tres categorias

```python
plt.figure(figsize=(10, 5))

plt.plot(temperatura_para_grafica, categoria_baja, label = "Baja", linewidth = 2)
plt.plot(temperatura_para_grafica, categoria_media, label = "Media", linewidth = 2)
plt.plot(temperatura_para_grafica, categoria_alta, label = "Alta", linewidth = 2)

plt.axvline(temperatura, linestyle="--", color="black", label="Dato: 35 grados")
plt.xlabel("Temperatura (grados Celsius)")
plt.ylabel("Grado de pertenencia (0 a 1)")
plt.title("Tres conjuntos difusos para una sola variable")
plt.ylim(-0.05, 1.05)
plt.grid(alpha=0.3)
plt.legend()
plt.show()
```
